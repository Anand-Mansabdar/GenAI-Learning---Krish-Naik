```python
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
GROQ_API_KEY = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(model="llama-3.1-8b-instant", api_key=GROQ_API_KEY)
llm
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

messages = [
  SystemMessage(content="Translate the following statement from English to German"),
  HumanMessage(content="Hello. Good Morning. How are you?")
]

print(llm.invoke(messages))
print(llm.invoke(messages).content)
```

```python
from langchain_core.output_parsers import StrOutputParser

result = llm.invoke(messages)
parser = StrOutputParser()
parser.invoke(result)
```

```python
# Using LCEL - Chaining the components using pipe(|)

chain = llm | parser
chain.invoke(messages)
```

```python
# Prompt Templates
from langchain_core.prompts import ChatPromptTemplate

template = "Translate the following into {language}:"

prompt_template = ChatPromptTemplate.from_messages(
  [
    ("system", template),
    ("user", "{text}")
  ]
)

result = prompt_template.invoke({"language": "German", "text": "Can I have a cup of coffee, please."})
```

```python
result.to_messages()
```

```python
chain = prompt_template | llm | parser
chain.invoke({"language": "German", "text": "Can I have a cup of coffee, please."})
```
