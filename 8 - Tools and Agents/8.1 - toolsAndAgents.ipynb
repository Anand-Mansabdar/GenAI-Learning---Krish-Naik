```python
# Creating In-built Tools
from langchain_community.tools import WikipediaQueryRun, ArxivQueryRun  # To run queries
from langchain_community.utilities import WikipediaAPIWrapper, ArxivAPIWrapper  # to conduct searches and fetch page summaries

from dotenv import load_dotenv

load_dotenv()
```

```python
# Wikipedia API Wrapper
wikipedia_api_wrapper = WikipediaAPIWrapper(top_k_results=2, doc_content_chars_max=150)

# Arxiv API Wrapper
arxiv_api_wrapper = ArxivAPIWrapper(top_k_results=3, doc_content_chars_max=250)
```

```python
# Query Run for Arxiv and Wikipedia
wikipedia_query_run = WikipediaQueryRun(api_wrapper=wikipedia_api_wrapper)

arxiv_query_run = ArxivQueryRun(api_wrapper=arxiv_api_wrapper)
```

```python
tools = [wikipedia_query_run, arxiv_query_run]
```

```python
# Creating Custom Tools
from langchain_community.document_loaders import WebBaseLoader
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
embeddings = HuggingFaceEmbeddings(
  model_name="BAAI/bge-small-en-v1.5"
)
```

```python
loader = WebBaseLoader("https://en.wikipedia.org/wiki/Virat_Kohli")
docs = loader.load()

documents = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200).split_documents(documents=docs)

vector_store = FAISS.from_documents(documents=documents, embedding=HuggingFaceEmbeddings())

retriever = vector_store.as_retriever()

retriever
```

```python
from langchain_classic.tools.retriever import create_retriever_tool

retriever_tool = create_retriever_tool(retriever, "search-virat-kohli", "Search about Virat Kohli on Wikipedia")

retriever_tool.name
retriever_tool.description
```

```python
tools.append(retriever_tool)

# print(tools)

for tool in tools:
  print(tool)
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(model="llama-3.3-70b-versatile")
```

```python
from langchain_classic import hub

prompt_template = hub.pull("hwchase17/openai-functions-agent")

prompt_template.messages
```

```python
from langchain_classic.agents import AgentExecutor
from langchain_classic.agents import create_openai_tools_agent

agent = create_openai_tools_agent(llm, tools, prompt=prompt_template)

agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)
```

```python
agent_executor.invoke({"input": "Tell me about Virat Kohli"})
```
