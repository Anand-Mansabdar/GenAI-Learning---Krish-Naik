```python
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

# Model Initialization
# Method 1:- without model_provider
model = init_chat_model(
  model="groq:llama-3.1-8b-instant",
)

# Method 2:- with model_provider
model = init_chat_model(
  model="llama-3.1-8b-instant",
  model_provider="GROQ"
)
```

```python
response = model.invoke("Explain cricket in short terms")
response.content
```

```python
# Initializing gemini_model
gemini_model = init_chat_model(
  model="google_genai:gemini-2.5-flash-lite"
)

response = gemini_model.invoke("Who is Virat Kohli. Explain in one word")
response.content
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

gemini_model = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
response = gemini_model.invoke("Who is Virat Kohli. Explain in one word")
response.content
```

```python
# Streaming

for chunk in model.stream("Write a paragraph on cricket."):
  print(chunk.text, end="")
```

```python
# BATCH

responses = model.batch([
  "What is cricket?",
  "Who is Virat Kohli?",
  "What did Cristinao Ronaldo achieve?"
])

for response in responses:
  print(response.content, end="")
```

```python
responses = model.batch([
  "What is cricket?",
  "Who is Virat Kohli?",
  "What did Cristinao Ronaldo achieve?"
], 
  config={
    "max_concurrency": 5,
  }
)

for response in responses:
  print(response.content, end="")
```
