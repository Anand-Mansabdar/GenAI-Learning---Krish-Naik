```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model(
  model="llama-3.1-8b-instant",
  model_provider="GROQ"
)

response = model.invoke("WHo is Cristinao Ronaldo in one word")
response.content
```

```python
from langchain.tools import tool

@tool
def get_weather(city: str) -> str:
  """To get weather of a particular city"""
  return f"The weather in {city} is cloudy"


model_with_tools = model.bind_tools([get_weather])
```

```python
response = model_with_tools.invoke("What is weather in Delhi?")
print(response)

for tool_call in response.tool_calls:
  print(f"Tool: {tool_call["name"]}")
  print(f"Arguments: {tool_call["args"]}")
```

```python
# TOOL EXECUTION LOOPS

messages = [{"role": "user", "content": "What's the weather in Delhi?"}]
ai_message = model_with_tools.invoke(messages)
messages.append(ai_message)


for tool_call in ai_message.tool_calls:
  tool_result = get_weather.invoke(tool_call)
  messages.append(tool_result)
  
final_response = model_with_tools.invoke(messages)
print(final_response.content)
```
