```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

from langchain.chat_models import init_chat_model

model = init_chat_model(
  model="llama-3.1-8b-instant",
  model_provider="GROQ"
)
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage, ToolMessage

messages = [
  SystemMessage("You are a poetry expert"),
  HumanMessage("Write a poem on Virat Kohli and Royal Challengers Bengaluru"),
]

response = model.invoke(messages)
print(response.content)
```

```python
# Passing Detailed Information to the LLM through the SystemMessage

messages = [
  SystemMessage("You are a senior Python developer with expertise in web frameworks. Always provide code examples and explain your reasoning. Be concise but thorough in your explainations."),
  HumanMessage("How to create a REST API")
]

response = model.invoke(messages)
print(response.content)
```

```python
human_message = HumanMessage(
  content="Hello",
  name="Anand Mansabdar",
  id="user1"
)
```

```python
response = model.invoke([human_message])
print(response.content)
```

```python
ai_message = AIMessage("Yes, sure! I would love to answer this question")

messages = [
  SystemMessage("You are a helpful message"),
  HumanMessage("Can you help me?"),
  ai_message,
  HumanMessage("Who is Virat Kohli? Explain in layman terms")
]

response = model.invoke(messages)
print(response)
```
