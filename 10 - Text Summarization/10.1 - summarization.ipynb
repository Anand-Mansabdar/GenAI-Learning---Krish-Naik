#### Prompt Template Text Summarization

```python
from langchain_classic.chains import LLMChain
from langchain_classic import PromptTemplate

generic_template = """
  Write a summary for the following: {paragraph}
  Translate the summary to {language}
"""

prompt = PromptTemplate(
  input_variables = ["speech", "language"],
  template=generic_template
)

prompt
```

```python
final_prompt = prompt.format(paragraph=paragraph, language="hindi")
```

```python
llm.get_num_tokens(final_prompt)
```

```python
llm_chain = LLMChain(llm=llm, prompt=prompt)
response = llm_chain.run({"paragraph":paragraph, "language": "hindi"})
print(response)
```
