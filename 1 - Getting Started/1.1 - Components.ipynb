```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ["MISTRAL_API_KEY"] = os.getenv("MISTRAL_API_KEY")

# The below keys are for langsmith
os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_PROJECT"] = os.getenv("LANGCHAIN_PROJECT")
```

```python
from langchain_mistralai import ChatMistralAI

llm = ChatMistralAI() # There is no need to write api_key attribute as it is already set in the environment variable
print(llm)
```

```python
result = llm.invoke("What is Cricket?")
print(result.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
  ("system", "You are an expert AI Enginner. Explain the asked question such that even a five-year old can understand"),
  ("user", "{question}") 
])

prompt
```

```python
chain = prompt | llm
response = chain.invoke({"question": "What is GenAI?"})
```

```python
print(response.content)
```

```python
# Output Parsers
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
chain = prompt | llm | parser
response = chain.invoke({"question": "What is GenAI?"})
```

```python
print(response)
```
